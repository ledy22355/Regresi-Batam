#OLS Regression Results

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm

data = {
    'Tahun'       : np.arange(2014, 2025),
    'BPendidikan' : (292.76, 286.07, 168.55, 265.51, 261.14, 129.78, 257.23, 59.25, 199.07, 107.50, 80.14),
    'BKesehatan'  : (243.90, 291.12, 100.94, 180.83, 121.78, 248.21, 194.40, 208.65, 249.48, 148.99, 278.77),
    'Kemiskinan'  : (10.48, 11.89, 8.94, 16.22, 11.31, 17.65, 15.42, 11.68, 13.28, 9.05, 10.91)}

df = pd.DataFrame(data)

x = df[['BPendidikan', 'BKesehatan']]
y = df['Kemiskinan']

x = sm.add_constant(x)

model = sm.OLS(y, x) # Menambahkan kolom konstanta untuk intercept
results = model.fit() # Membangun model regresi (Ordinary Least Square)

print(results.summary())
```

#Regresi Linear Berganda 3D

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
import plotly.graph_objects as go

# DATA KOTA BATAM (2014-2024)
data = {
    "Tahun"            : np.arange(2014, 2025),
    "Biaya Pendidikan" : [292.76, 286.07, 168.55, 265.51, 261.14, 129.78, 257.23, 59.25, 199.07, 107.50, 80.14],
    "Biaya Kesehatan"  : [243.90, 291.12, 100.94, 180.83, 121.78, 248.21, 194.40, 208.65, 249.48, 148.99, 278.77],
    "Kemiskinan"       : [10.48, 11.89, 8.94, 16.22, 11.31, 17.65, 15.42, 11.68, 13.28, 9.05, 10.91]}

df = pd.DataFrame(data)

# VARIABEL INDEPENDEN DAN DEPENDEN
X = df[["Biaya Pendidikan", "Biaya Kesehatan"]]
y = df["Kemiskinan"]

# MODEL OLS
X_model = sm.add_constant(X)
results = sm.OLS(y, X_model).fit()

intercept = results.params["const"]
coef_pendidikan = results.params["Biaya Pendidikan"]
coef_kesehatan = results.params["Biaya Kesehatan"]
r_squared = results.rsquared

persamaan = (
    f"Ŷ = {intercept:.3f}"
    f" + ({coef_pendidikan:.4f}) X₁"
    f" + ({coef_kesehatan:.4f}) X₂")

# GRID BIDANG REGRESI
x1 = np.linspace(df["Biaya Pendidikan"].min(), df["Biaya Pendidikan"].max(), 40)
x2 = np.linspace(df["Biaya Kesehatan"].min(), df["Biaya Kesehatan"].max(), 40)
X1, X2 = np.meshgrid(x1, x2)
Y = intercept + coef_pendidikan * X1 + coef_kesehatan * X2

# GRAFIK 3D
fig = go.Figure()

# Data aktual
fig.add_trace(
    go.Scatter3d(
        x=df["Biaya Pendidikan"],
        y=df["Biaya Kesehatan"],
        z=df["Kemiskinan"],
        mode="markers",
        marker=dict(size=7, color="red", opacity=0.9),
        name="Data Aktual",
        text=[f"Tahun {t}" for t in df["Tahun"]],
        hovertemplate=(
            "<b>%{text}</b><br>"
            "Biaya Pendidikan: %{x}<br>"
            "Biaya Kesehatan: %{y}<br>"
            "Kemiskinan: %{z}"
            "<extra></extra>")))

# Bidang regresi
fig.add_trace(
    go.Surface(
        x=X1, y=X2, z=Y,
        opacity=0.60,
        colorscale="Viridis",
        name="Regression Plane",
        hovertemplate=(
            "Biaya Pendidikan: %{x:.2f}<br>"
            "Biaya Kesehatan: %{y:.2f}<br>"
            "Prediksi Kemiskinan: %{z:.2f}"
            "<extra></extra>")))

fig.update_layout(
    title={
        "text": (
            "REGRESI LINEAR BERGANDA 3D - KOTA BATAM"
            "<br>"
            f"<sup>{persamaan} | R² = {r_squared:.3f}</sup>"),
        "x": 0.5},
    scene={
        "xaxis": {"title": "Biaya Pendidikan (X₁)"},
        "yaxis": {"title": "Biaya Kesehatan (X₂)"},
        "zaxis": {"title": "Kemiskinan (Y)"}},
    width=1000,
    height=750,
    margin={"l": 0, "r": 0, "b": 0, "t": 100},
    legend={"x": 0.02, "y": 0.98})

fig.show()
```
